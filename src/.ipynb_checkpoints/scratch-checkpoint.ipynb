```python
# try taking advantage of linear subtree isomorphism

# generate random trees
def randomTree(n):
    T = nx.Graph()
    T.add_node(0)
    
    for i in range(1, n):
        r = rnd.randint(0,i-1)
        T.add_edge(r, i)
    
    return T


# return the canonical string of a tree rooted at 'root'
def compressTree(T, node, parent):
    string = ''
    
    if(node == parent):
        children = set(T[parent])

    else:
        children = set(T[node]).difference({parent})
        
    if(children):
        for n in children:
            s = '1'
            s = s + compressTree(T, n, node)
            s = s + '0'
            string = string + s
    
    else:
        return ''

    return string


# match s to an ordered character sequence in t
def matchSequence(s, t):
    i = 0
    l = []
    
    for c in s:
        while(t[i] != c):
            i = i + 1
        print(t[i])
        l.append(t[i])
        
    if(l == s):
        print(l)
        return True
    
    print(l)
    return False


# test subtree isomorphism on T and S
def subtreeIsomorphism(T, S):
    pass
```
